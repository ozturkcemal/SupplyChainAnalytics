```python
import pyomo.environ as pyo

# Items that could be loaded into the 20 ft container
A = ['crate', 'coil', 'pallet', 'cartons']   # machinery crate, steel coil, pallet of tiles, electronics cartons
b = {'crate': 8, 'coil': 3, 'pallet': 6, 'cartons': 11}   # profit of each item (k€)
w = {'crate': 5, 'coil': 7, 'pallet': 4, 'cartons': 3}    # weight of each item (tonnes)

W_max = 14   # container payload capacity (tonnes)

model = pyo.ConcreteModel()
model.x = pyo.Var( A, within=pyo.Binary )

model.obj = pyo.Objective(
    expr = sum( b[i]*model.x[i] for i in A ),
    sense = pyo.maximize )

model.weight_con = pyo.Constraint(
    expr = sum( w[i]*model.x[i] for i in A ) <= W_max )

#opt = pyo.SolverFactory('glpk')
opt = pyo.SolverFactory('cbc')
opt_success = opt.solve(model)

model.pprint()
# Print the values of the decision variables
print("\nSelected items:")
for i in A:
    print(f"{i}: {model.x[i].value}")

print(f"Objective: {model.obj()}")
```
